```python
#• Challenge 1: Write a program that asks a user for their monthly income and monthly expenses using input(). Calculate their savings rate percentage. If the savings rate is above 20%, print "Healthy financial habits!". If it is between 10% and 20%, print "Decent, but room to optimize." Otherwise, print "Warning: High spending risk."
I = int(input("Enter monthly income"))
E = int(input("Enter monthly expenses"))
savings = I-E
rate = (savings/I)*100
if rate>20:
  print("Healthy financial habits!")
elif rate>=10:
  print("Decent, but room to optimize")
else:
  print("Warning: High spending risk")
```
